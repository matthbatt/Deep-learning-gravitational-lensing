## Initial grid

```python
import numpy as np
N = 40
grid = np.ones([N,N])
mask = np.random.random((N, N)) < 0.5
# Flip those sites
grid[mask] = -1
```

```python
import matplotlib.pyplot as plt

plt.imshow(grid, cmap="gray")
plt.axis("off")
plt.show()
```

```python
from scipy.ndimage import convolve
from scipy.ndimage import generate_binary_structure

def hamilton(lattice):
    kern = generate_binary_structure(2, 1)
    kern[1, 1] = False

    arr = -lattice * convolve(
        lattice,
        kern,
        mode='constant',
        cval=0
    )

    return arr.sum() / 2
```

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

beta = 1

fig, ax = plt.subplots(figsize=(6, 6))

im = ax.imshow(grid, cmap="gray", vmin=-1, vmax=1)
ax.axis("off")
im.set_data(grid)

def update(frame):
    global grid

    # Perform 100 Metropolis updates
    for _ in range(2000):
        i = np.random.randint(N)
        j = np.random.randint(N)

        grid_2 = grid.copy()
        grid_2[i, j] *= -1

        dH = hamilton(grid_2) - hamilton(grid)

        if dH <= 0 or np.random.random() < np.exp(-beta * dH):
            grid = grid_2

    # Display only after the 100 updates
    im.set_data(grid)
    ax.set_title(f"Iteration {100 * (frame + 1)}")

    return [im]

ani = FuncAnimation(
    fig,
    update,
    frames= 1000,
    interval=20,
    blit=True
)

HTML(ani.to_jshtml())
```
