```python
from scipy.stats import norm
from math import log, sqrt, exp
from scipy.optimize import brentq

def bs_call(S, K, T, r, sigma):
    d1 = (log(S/K) + (r + 0.5*sigma**2)*T) / (sigma*sqrt(T))
    d2 = d1 - sigma*sqrt(T)
    return S*norm.cdf(d1) - K*exp(-r*T)*norm.cdf(d2)

def implied_vol_call(price, S, K, T, r):
    f = lambda sigma: bs_call(S, K, T, r, sigma) - price
    return brentq(f, 1e-6, 5.0)
```

```python
import yfinance as yf

ticker = yf.Ticker("AAPL")
expirations = ticker.options   # list of available maturities
```

```python
import pandas as pd

chains = {}

for expi in expirations:
    opt = ticker.option_chain(expi)
    calls = opt.calls
    puts = opt.puts
    chains[expi] = pd.concat([calls, puts], ignore_index=True)
```

```python
import numpy as np
import pandas as pd

surface = []

S = ticker.history(period="1d")["Close"].iloc[-1]
r = 0.02  # risk-free rate

for expi in expirations:
    T = (pd.to_datetime(expi) - pd.Timestamp.today()).days / 365
    df = chains[expi]

    for _, row in df.iterrows():
        K = row["strike"]
        bid = row["bid"]
        ask = row["ask"]
        mid = 0.5 * (bid + ask)

        # -----------------------------
        # 1. Skip garbage quotes
        # -----------------------------
        if mid <= 0 or (bid == 0 and ask == 0):
            continue

        # 2. Skip tiny midprices (deep OTM noise)
        if mid < 0.05:
            continue

        # 3. No-arbitrage check for CALLS
        intrinsic = max(S - K, 0)
        if mid < intrinsic:
            continue

        # -----------------------------
        # 4. Compute implied volatility
        # -----------------------------
        try:
            iv = implied_vol_call(mid, S, K, T, r)
            surface.append([K, T, iv])

        except ValueError:
            # Brent fails when f(a) and f(b) have same sign
            continue

        except Exception as e:
            print("Unexpected IV error:", e, "K=", K, "T=", T, "mid=", mid)
            continue

# Convert to DataFrame
surface = pd.DataFrame(surface, columns=["Strike", "Maturity", "IV"])
```

```python
import numpy as np
from scipy.interpolate import griddata
import matplotlib.pyplot as plt

# Use your downsampled points
X = sampled["Strike"].values
Y = sampled["Maturity"].values
Z = sampled["IV"].values



# Create a regular grid
strike_grid = np.linspace(X.min(), X.max(), 15)
maturity_grid = np.linspace(Y.min(), Y.max(), 15)

Sg, Tg = np.meshgrid(strike_grid, maturity_grid)

# Interpolate IV onto the grid
Zg = griddata((X, Y), Z, (Sg, Tg), method='linear')
```

```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams["text.usetex"] = True
plt.rcParams["font.family"] = "serif"
plt.rcParams["font.serif"] = ["Computer Modern Roman"]

# ----------------------------------------------------
# Generate example volatility smile points (V-shape)
# ----------------------------------------------------
np.random.seed(0)

strikes = np.linspace(80, 120, 20)

# V-shaped smile using absolute value
true_smile = -1.0 + 2.14 * np.abs(strikes - 100) + 0.08*(strikes - 100)**2 - 1.2*(strikes - 100)

# Add some noise to simulate market data
noise = np.random.normal(0, 6, size=strikes.size)
iv_market = true_smile + noise

# ----------------------------------------------------
# Fit a V-shape using least squares on |K - K0|
# ----------------------------------------------------
K0 = 100  # center strike
X = np.abs(strikes - K0)
A = np.vstack([np.ones_like(X), X]).T
coef = np.linalg.lstsq(A, iv_market, rcond=None)[0]

a, b = coef
strikes_fit = np.linspace(strikes.min(), strikes.max(), 300)
iv_fit = a + b * np.abs(strikes_fit - K0) + 0.05*(strikes_fit - 100)**2 - 1.7*(strikes_fit - 100)

# ----------------------------------------------------
# Plot the smile
# ----------------------------------------------------
plt.figure(figsize=(8,4))

plt.scatter(strikes, iv_market, color="black", label="$\sigma$ du marché", s=40)
plt.plot(strikes_fit, iv_fit, color="red", linewidth=2, label="Modèle de volatilité")

plt.xlabel("Strike K", fontsize  = 16)
plt.ylabel("Volatility", fontsize  = 16)
plt.title("Modélisation de la volatilité", fontsize  = 16)
plt.grid(True)
plt.legend()
plt.savefig("/net/CLUSTER/VAULT/users/mbatt/gravitational_lensing/plots/vol_interpol.png", dpi=300, transparent=True)
plt.show()
```
