```python
import yfinance as yf
import numpy as np
import matplotlib.pyplot as plt

# Enable LaTeX style
plt.rcParams["text.usetex"] = True
plt.rcParams["font.family"] = "serif"
plt.rcParams["font.serif"] = ["Computer Modern Roman"]

# -----------------------------
# 1. Download S&P 500 data
# -----------------------------
ticker = "^GSPC"
data = yf.download(ticker, period="5y")
prices = data["Close"].dropna()

n = len(prices)

# -----------------------------
# 2. Brownian motion with drift
# -----------------------------
dt = 1
mu = 0.7

W = np.zeros(n)
W[1:] = mu + np.cumsum(np.sqrt(dt) * np.random.randn(n - 1))

price_min = float(prices.min())
price_max = float(prices.max())

W_scaled = (W - W.min()) / (W.max() - W.min())
W_scaled = W_scaled * (price_max - price_min) + price_min

# Drift line
drift_line = np.linspace(price_min, price_max, n)

# -----------------------------
# 3. Plot
# -----------------------------
plt.figure(figsize=(8, 6))

plt.plot(prices.index, prices.values, label=r"S\&P 500", linewidth=2)
plt.plot(prices.index, W_scaled, label=r"Mouvement Brownien avec dérive", alpha=0.8)
# plt.plot(prices.index, drift_line, "--", color="red", label=r"Drift")

plt.title(r"S\&P 500 vs Mouvement Brownien avec dérive", fontsize=16)
plt.xlabel(r"Date", fontsize=14)
plt.ylabel(r"Valeur", fontsize=14)
plt.grid(True)
plt.legend(fontsize=12)
plt.savefig("/net/CLUSTER/VAULT/users/mbatt/gravitational_lensing/plots/bm_sp500.png", dpi=300, bbox_inches="tight", transparent=True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

plt.rcParams["text.usetex"] = True
plt.rcParams["font.family"] = "serif"
plt.rcParams["font.serif"] = ["Computer Modern Roman"]

# -----------------------------
# Parameters
# -----------------------------
T = 1.0
N = 250
dt = T / N
t = np.linspace(0, T, N)

S0 = 100
mu = 0.05
sigma = 0.02
strike = 107
today = 0.6
today_idx = int(today * N)

# -----------------------------
# Past Brownian motion (seeded)
# -----------------------------
np.random.seed(123)
W_past = np.cumsum(np.sqrt(dt) * np.random.randn(today_idx))
S_past = S0 * np.exp((mu - 0.5*sigma**2)*t[:today_idx] + sigma*W_past)
S_today = S_past[-1]

# -----------------------------
# Future simulations (NO seed)
# -----------------------------
n_paths = 200
future_steps = N - today_idx

W_future = np.cumsum(
    np.sqrt(dt) * np.random.randn(n_paths, future_steps),
    axis=1
)

S_future = S_today * np.exp(
    (mu - 0.5*sigma**2)*(t[today_idx:] - today) + sigma*W_future
)

S_T = S_future[:, -1]   # distribution at maturity

# -----------------------------
# Create two subplots sharing Y axis
# -----------------------------
fig, (ax1, ax2) = plt.subplots(
    1, 2, sharey=True, figsize=(12, 6),
    gridspec_kw={"width_ratios": [3, 1]}
)

# -----------------------------
# Left plot: Brownian motion
# -----------------------------
ax1.plot(t[:today_idx], S_past, linewidth=2, label="Past path")
ax1.grid(True)

for i in range(80):
    ax1.plot(t[today_idx:], S_future[i], alpha=1)

ax1.axvline(today, color="black", linestyle="--", label="Today")
ax1.axhline(strike, color="red", linewidth=2)

ax1.set_xlabel("Time")
ax1.set_ylabel("Price")
ax1.set_title("Brownian Motion with Future Simulations")
ax1.legend()

# -----------------------------
# Right plot: Rotated histogram
# -----------------------------
ax2.hist(
    S_T,
    bins=30,
    density=True,
    alpha=0.6,
    color="skyblue",
    orientation="horizontal",
    edgecolor="black"
)

# Bell curve (horizontal)
mean = np.mean(S_T)
std = np.std(S_T)
y_vals = np.linspace(mean - 4*std, mean + 4*std, 200)
x_vals = norm.pdf(y_vals, mean, std)

ax2.plot(x_vals, y_vals, color="darkblue", linewidth=2)

ax2.set_xlabel("Density")
ax2.set_title("Distribution at Maturity")

plt.tight_layout()
plt.show()
```
